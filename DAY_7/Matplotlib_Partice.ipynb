## 1. Line Chart

```python
import matplotlib.pyplot as plt
Sales=[200, 250, 300, 280, 350]
Year=[2019, 2020, 2021, 2022, 2023]
plt.plot(Year,Sales)
plt.title("Line Chart")
plt.ylabel('SALES')
plt.xlabel('YEAR')
plt.show()
```

## 2. Bar Chart

```python
import matplotlib.pyplot as plt
subjects = ['Math', 'Science', 'English', 'History']
marks = [85, 92, 78, 88]
plt.bar(subjects, marks)
plt.title("Bar Chart")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

## 3. Histogram

```python
import matplotlib.pyplot as plt
ages = [18, 19, 20, 20, 21, 22, 22, 23, 24, 25, 25, 26, 27, 28, 29, 30, 30, 31]
plt.hist(ages, bins=6,color='red')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

## 4. Scatter Plot

```python
import matplotlib.pyplot as plt
hours_studied = [1, 2, 3, 4, 5, 6]
scores = [40, 50, 60, 70, 85, 95]
plt.scatter(hours_studied, scores)
plt.title("Study Hours vs Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.show()
```

## 5. Pie Chart

```python
import matplotlib.pyplot as plt
expenses = [5000, 3000, 2000, 1000]
categories = ['Rent', 'Food', 'Travel', 'Others']
plt.pie(expenses, labels=categories)
plt.title("Monthly Expenses")
plt.show()
```

## 6. Box Plot

```python
import matplotlib.pyplot as plt
exam_scores = [55, 60, 65, 70, 72, 75, 80, 85, 90, 95]
plt.boxplot(exam_scores)
plt.title("Exam Scores Distribution")
plt.ylabel("Scores")
plt.show()
```

## 7. Heatmap

```python
import matplotlib.pyplot as plt
import numpy as np
data = np.array([
    [5, 8, 7],
    [6, 9, 3],
    [4, 2, 8]
])
plt.imshow(data, cmap='hot')
plt.colorbar()
plt.title("Heatmap Example")
plt.show()
```

## Customizing 

## 1. Customizing Line Chart

```python
import matplotlib.pyplot as plt
Sales=[200, 250, 300, 280, 350]
Year=[2019, 2020, 2021, 2022, 2023]
plt.plot(Year,Sales,color='orange',linewidth=4,marker='o',markersize=10,linestyle='--')
plt.title("Line Chart")
plt.ylabel('SALES')
plt.xlabel('YEAR')
plt.show()
```

## 2. Customizing Bar Chart

```python
import matplotlib.pyplot as plt
subjects = ['Math', 'Science', 'English', 'History']
marks = [85, 92, 78, 88]
plt.bar(subjects, marks, color='green', edgecolor='black', linewidth=2)
plt.title("Bar Chart")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

## 3. Customizing Histogram Plot

```python
import matplotlib.pyplot as plt
ages = [18, 19, 20, 20, 21, 22, 22, 23, 24, 25, 25, 26, 27, 28, 29, 30, 30, 31]
plt.hist(ages, bins=6,color='green', edgecolor='blue',linestyle='--', alpha=0.5)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

## 4. Customizing Scatter Plot

```python
import matplotlib.pyplot as plt
hours = [1, 2, 3, 4, 5, 6]
scores = [40, 50, 60, 70, 85, 95]
sizes = [100, 200, 300, 400, 500, 600]
colors = [10, 20, 30, 40, 50, 60]
plt.scatter(hours, scores,s=sizes,c=colors,marker='D',alpha=0.6)
plt.title("Hours vs Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Scores")
plt.show()
```

## 5. Customized Pie Chart

```python
import matplotlib.pyplot as plt
expenses = [40, 25, 20, 15]
labels = ['Food', 'Rent', 'Travel', 'Other']
explode = [0.1, 0, 0, 0]
colors = ['gold', 'lightblue', 'lightgreen', 'pink']
plt.pie(expenses,labels=labels,colors=colors,explode=explode,autopct='%1.1f%%',shadow=True)
plt.title("Monthly Expenses")
plt.show()
```

## 1. Figure class

```python
import matplotlib.pyplot as plt
fig = plt.figure(figsize=(7, 5), facecolor='lightyellow')
ax = fig.add_axes([0.15, 0.15, 0.75, 0.75])
subjects = ['Math', 'Science', 'English', 'Social']
marks = [85, 92, 78, 88]
ax.plot(subjects, marks,color='green',marker='o',linewidth=2)
ax.set_title("Student Marks")
ax.set_xlabel("Subjects")
ax.set_ylabel("Marks")
plt.show()
```

## 2. Axes Class

```python
import matplotlib.pyplot as plt
students = [1, 2, 3, 4, 5]
marks = [70, 85, 90, 75, 95]
attendance = [80, 88, 92, 85, 98]
fig = plt.figure(figsize=(6, 4))
ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])
ax.plot(students, marks, marker='o')
ax.plot(students, attendance, marker='s')
ax.set_title("Marks vs Attendance")
ax.set_xlabel("Student Number")
ax.set_ylabel("Percentage")
ax.legend(["Marks", "Attendance"])
plt.show()
```

## Advanced Techniques for Visualizing Subplots

## 1. Using add_axes()

```python
import matplotlib.pyplot as plt
x=[20,30,50,60]
y=[30,40,55,75]
fig = plt.figure(figsize=(10, 4)) 
ax1 = fig.add_axes([0.1, 0.1, 0.35, 0.8]) 
ax2 = fig.add_axes([0.55, 0.1, 0.35, 0.8])
ax1.plot(x, y)
ax2.plot(y, x)
plt.show()
```

## 2. Using subplot()

```python
import matplotlib.pyplot as plt
x=[10,25,35,50,60]
y=[25,40,55,60,70]
plt.figure()
plt.subplot(121)
plt.plot(x, y)
plt.subplot(122)
plt.plot(y, x)
plt.show()
```

## 3. Using subplot2grid()

```python
import matplotlib.pyplot as plt
x = [10,25,35,50,60]
y = [20,30,45,65,75]
axes1 = plt.subplot2grid((7, 1), (0, 0), rowspan = 2, colspan = 1)
axes2 = plt.subplot2grid ((7, 1), (2, 0), rowspan = 2, colspan = 1)
axes1.plot(x, y)
axes2.plot(y, x)
plt.show()
```

## Saving Plots Using savefig()

```python
import matplotlib.pyplot as plt
year = ['2020', '2021', '2022', '2023', '2024']
production = [150, 130,200,230,270]
plt.bar(year, production)
plt.savefig("output.jpg")
plt.savefig("output1", facecolor='y', bbox_inches="tight",pad_inches=0.3, transparent=True)
```
