## Seaborn

## 1. Line plot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.DataFrame({
    'Day': [1, 2, 3, 4, 5],
    'Sales': [200, 250, 300, 280, 350]
})
sns.lineplot(x='Day', y='Sales', data=df)
plt.title("Daily Sales Trend")
plt.show()
```

## 2. Scatter Plot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.DataFrame({
    'Hours_Studied': [2, 4, 6, 8, 10],
    'Marks': [40, 50, 65, 80, 95]
})
sns.scatterplot(x='Hours_Studied', y='Marks', data=df)
plt.title("Study Hours vs Marks")
plt.show()
```

## 3. Box plot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.DataFrame({
    'Salary': [25000, 27000, 29000, 31000, 33000, 50000]
})
sns.boxplot(y='Salary', data=df)
plt.title("Salary Distribution")
plt.show()
```

## 4. Violin Plot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.DataFrame({
    'Department': ['IT', 'IT', 'HR', 'HR', 'Sales', 'Sales'],
    'Salary': [40000, 45000, 35000, 38000, 30000, 32000]
})
sns.violinplot(x='Department', y='Salary', data=df)
plt.title("Department Salary Distribution")
plt.show()
```

## 5. Swarm Plot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.DataFrame({
    'Class': ['A', 'A', 'A', 'B', 'B', 'B'],
    'Marks': [75, 80, 85, 70, 78, 88]
})
sns.swarmplot(x='Class', y='Marks', data=df)
plt.title("Marks Distribution")
plt.show()
```

## 6. Bar Plot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.DataFrame({
    'Product': ['Laptop', 'Mobile', 'Tablet'],
    'Sales': [120, 200, 90]
})
sns.barplot(x='Product', y='Sales', data=df)
plt.title("Product Sales")
plt.show()
```

## 7. Point Plot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar', 'Apr'],
    'Profit': [20, 35, 30, 50]
})
sns.pointplot(x='Month', y='Profit', data=df)
plt.title("Monthly Profit")
plt.show()
```

## 8. Count Plot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Male', 'Male', 'Female']
})
sns.countplot(x='Gender', data=df)
plt.title("Gender Count")
plt.show()
```

## 9. KDE Plot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
df = pd.DataFrame({
    'Marks': np.random.normal(70, 10, 100)
})
sns.kdeplot(data=df, x='Marks', fill=True)
plt.title("Marks Density Distribution")
plt.show()
```

## Customizing Seaborn plots

## 1. Adding Titles and Axis Labels

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
df = pd.DataFrame({
    'Hours': [1, 2, 3, 4, 5],
    'Marks': [35, 45, 60, 75, 90]
})
sns.scatterplot(x='Hours', y='Marks', data=df)
plt.title("Study Hours vs Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks Obtained")
plt.show()
```

## 2. Built-in Styles and Grids in Seaborn

```python
from sklearn.datasets import load_iris
import seaborn as sns
import matplotlib.pyplot as plt
iris = load_iris()
sns.set_style("whitegrid")

sns.boxplot(x='species', y='petal_length', data=sns.load_dataset('iris'),color='red')
plt.title('Petal Length Distribution by Species')
plt.show()

```

## 3. Customizing Color Palettes

a) Using a Built-in Palette:

```python
sns.set_palette("pastel") 
sns.violinplot(x='species', y='petal_length', data=sns.load_dataset('iris'))
plt.title('Petal Length Distribution by Species')
plt.show()
```

b) Using a Custom Palette:

```python
custom_colors = ['#FF5733', '#33FFBD', '#335BFF']
sns.set_palette(custom_colors)
sns.violinplot(x='species', y='petal_length', data=sns.load_dataset('iris'))
plt.title('Custom Colored Petal Length Distribution')
plt.show()
```

## 4. Adjusting Figure Size and Aspect Ratio

```python
plt.figure(figsize=(10, 6))
sns.lineplot(x='year', y='passengers', data=sns.load_dataset('flights'))
plt.title('Number of Passengers Over Time')
plt.show()
```

## 5. Adding Markers to Line Plots

```python
sns.lineplot(x='year', y='passengers', data=sns.load_dataset('flights'), marker='o')
plt.title('Number of Passengers Over Time')
plt.show()
```

## Visualizing Relationships and Patterns with Seaborn

## 1. Pair Plots

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_style("whitegrid")
custom_palette = sns.color_palette("husl", 8)
sns.set_palette(custom_palette)
data = sns.load_dataset("iris")
sns.pairplot(data, hue="species")
plt.show()
```

## 2. Joint Plots

```python
import seaborn as sns
import matplotlib.pyplot as plt
data = sns.load_dataset("tips")
sns.jointplot(x="total_bill", y="tip", data=data, kind="scatter", color="#008B8B")
plt.show()
```

## 3. Grid Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips")
plot=sns.FacetGrid(tips, col="time", row="sex")
plot.map(sns.scatterplot, "total_bill", "tip")
plt.show()
```

## Regression Plots: Visualizing Linear Relationships

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset('tips')
sns.regplot(x='total_bill', y='tip', data=tips, scatter_kws={'s':10}, line_kws={'color':'red'})
plt.show()
```
