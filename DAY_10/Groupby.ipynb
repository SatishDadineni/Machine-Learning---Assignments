## 1. GROUPBY Operations

```python
import pandas as pd

df = pd.read_csv("customer_shopping_behavior.csv")

# Average purchase amount by gender
print("\n",df.groupby("Gender")["Purchase Amount (USD)"].mean())

# Total purchase amount by category
print("\n",df.groupby("Category")["Purchase Amount (USD)"].sum())

# Average rating by category
print("\n",df.groupby("Category")["Review Rating"].mean())

# Number of customers by subscription status
print("\n",df.groupby("Subscription Status")["Customer ID"].count())

# Total purchases by payment method
print(df.groupby("Payment Method")["Purchase Amount (USD)"].sum())

# Multiple aggregations
print("\n")
print(
    df.groupby("Category").agg({
        "Purchase Amount (USD)": ["sum", "mean", "max"],
        "Review Rating": "mean"
    })
)
```

## 2. PIVOT Operations

```python
# Purchase Amount by Gender and Category
pivot1 = df.pivot_table(
    values="Purchase Amount (USD)",
    index="Gender",
    columns="Category",
    aggfunc="sum"
)

print("\n pivot1:\n",pivot1)

# Average Rating by Gender and Category
pivot2 = df.pivot_table(
    values="Review Rating",
    index="Gender",
    columns="Category",
    aggfunc="mean"
)

print("\n pivot2:\n",pivot2)

# Count of Customers by Gender and Season
pivot3 = df.pivot_table(
    values="Customer ID",
    index="Gender",
    columns="Season",
    aggfunc="count"
)
print("\n pivot3:\n",pivot3)
```

## 3. MERGE Operations

```python
# Customer details table
customer_df = df[["Customer ID", "Age", "Gender", "Location"]]

# Purchase details table
purchase_df = df[["Customer ID", "Item Purchased",
                  "Category", "Purchase Amount (USD)"]]

# Inner Merge
merged_df = pd.merge(
    customer_df,
    purchase_df,
    on="Customer ID",
    how="inner"
)

print(merged_df.head())
```
